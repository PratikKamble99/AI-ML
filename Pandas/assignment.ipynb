```python
df = pd.read_csv("IRIS.csv")

# Q1. A- Display Following Information - First 10 rows, shape & data type, Summery Statistic
df.head(10)
print(df.shape)
print(df.dtypes)

df.describe()
```

```python
# B. Select only those rows where - petal_length > 4.5, species="lris-virginica"
df[(df["petal_length"] > 4.5) & (df["species"]=="Iris-virginica")] # always use () when appling two contions with & or |
```

```python
# Group by species and compute
df.groupby("species").agg(
    avg_sepal_length=("sepal_length","mean"), 
    max_petal_width=("petal_width","max"), 
    std_sepal_width=("sepal_width","std")
)
```

```python
# create new column petal_ratio = petal_length/petal_width - For each species, find the average petal_ratio
df["petal_ratio"] = df["petal_length"] / df["petal_width"]

df.groupby("species")["petal_ratio"].mean()
```
