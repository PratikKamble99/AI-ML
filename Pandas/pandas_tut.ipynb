
# Data types

```python
# Series - 1D lablled array
'''  Named Index - 
     labels are the index values assigned to each data point, 
     which serve as unique identifiers for data retrieval and alignment

     By default, if no custom labels are specified, 
     integer positions (starting from 0) are used as labels. 
'''
s = pd.Series([1, 2, 3, 4, 5])

print(s)

# Index
print(s[0]) # index based access
```

```python
# index argument used to pass custom lables in series
s = pd.Series([1, 2, 3, 4, 5], index=["a","b", "c", "d", "e"]) 
print(s)
print(s["a"]) # labled based access
print(s.index)
```

```python
# Series Properties
1. Homogenious
2. Vectorization
3. Handle issing values with NAN
4. Mutable values, immutable size
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

# vectorization
print(s1+s2)

#Mutable values
s1[0] = 100

# Immutable Size
changed_s1 = s1.drop(0) # This create new array does not change original
print(s1)
print(changed_s1)
```

# Selecting Data

```python
# By Column
print(df["city"])
print(df[["city","aqi"]])
```

```python
# By Row -  lable based (loc)
#        -  Index Based (iloc)

df.loc[0] # 0 lable rows data
df.loc[0:3] # from 0 to 3 lable rows data - ending is inclusive

df.iloc[0] # 0 index rows data
df.iloc[0:3] # from 0 to 2 index rows data - ending is exlusive
```

```python
# By both - row and column
df.loc[0:2, ["city", "aqi"]]

# this will throw error
# df.iloc[0:2, ["city", "aqi"]] # iloc expects both rows and columns index not value eg. ["city", "aqi"] this are column lable
df.iloc[0:3, [2, 9]] # [2, 9] are column index
df.iloc[0:3, 2:9] # 2:9 are column index range
```

```python
# select single scalar value - at & iat
df.at[0, "city"]
df.iat[0, 2] # i means index

# When we access any data using above functions it return view of data not copy
```

# Filtering Data


```python
df[ df["aqi"] > 100 ]
```

```python
# Operatores
# AND - &
# OR - |
# not - ~

df[ (df["aqi"] > 100) & (df["temperature"] > 30) ]

# Chaining
df[ (df["aqi"] > 100) & (df["temperature"] > 30) ][["city", "aqi"]]

# filtering data preserve lables
aqi_data = df[ (df["aqi"] > 100) & (df["temperature"] > 30) ][["city", "aqi"]]
aqi_data.iloc[0] # will return 1st record
# aqi_data.loc[0] # will throw error because lable not in filtered data
aqi_data.loc[6] 
```

```python
# filtering Data using query
# use backtics for special char
df.query("aqi > 100 & temperature > 30") # Pass exp as string always
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]

# pass your python variables
aqi_value = 100
df.query("aqi > @aqi_value & temperature > 30")[["city", "aqi"]]
```

# Cleaning Data

```python
# Cleaning data - missing values
df = pd.read_csv("raw_data.csv")
```

```python
# get data in True-False format
df.isnull()
df.isna()

df.isnull().sum() # get how many missing values count with key

df.dropna() # delete null value contain rows
df.dropna(axis=1) # drop column

df.fillna(0) # fill empty cells with value

age_mean = df["age"].mean()
df["age"].fillna(age_mean)

df
# forwardfill - fill current cell based on prev cell value 
df.ffill()
# backwardfill - fill current cell based on next cell value 
df.bfill()
```

```python
# Handle duplicate values
df.duplicated() # get in True False
df["country"].duplicated() 

# delete duplicates
df.drop_duplicates()
# df.drop_duplicates(inplace=True) # make change in original dataframe
```

```python
# Handle datatype and date-time
df.dtypes

df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] =df2["age"].astype("int64")
df2.dtypes
```
