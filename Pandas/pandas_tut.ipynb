# Transforming Data

```python
# feature Engineering

emp_data = df.copy()

# .apply() - It takes function as arguments which is our custom logic to transform data ie apply is higher order function
emp_data["tax"] = emp_data["income"].apply(lambda x: "20%" if x >= 50000 else "10%")

# .map()
gender = {"Male":"M", "Female":"F", "unknown":"U"}
emp_data["gender"] = emp_data["gender"].map(gender)

# .assign() - it craete new column by assigning new value
emp_data = emp_data.assign(new_income = emp_data["income"] * 1.1)

# .replace(old, new)
emp_data["country"] = emp_data["country"].replace("USA", "US")
emp_data
```

```python
# rename column names
# All columns
emp_data.columns = ["Id","Name",	"Age",	"Country",	"Gender",	"Income",	"Tax",	"New_income"]

# rename perticular col
emp_data = emp_data.rename(columns={"Income":"Salary"})
emp_data.rename(index={1:"First"})

# .sort_values()
emp_data.sort_values("Salary", ascending=False)
sorted_data = emp_data.sort_values(["Salary", "Age"])

sorted_data.sort_index() # get data in original form
sorted_data.reset_index() # reset only index not sorted data and this add new col index
sorted_data.reset_index(drop=True) # it resets original index with new index

# sorted_data
```

```python
# .rank()
sorted_data["ranking"] = sorted_data["Salary"].rank(ascending=False, method="dense") # dense set whole rank if ties
sorted_data["ranking"] = sorted_data["Salary"].rank(ascending=False, method="min") # min set min rank to both if ties
sorted_data["ranking"] = sorted_data["Salary"].rank(ascending=False, method="max") # max set max rank to both rank if ties

# reordering of columns
sorted_data[["Id","Name",	"Age",	"Country",	"Gender",	"Salary",	"New_income", "Tax",]]
```

# Grouping and aggrigation of Data

# Melt And pivot 

```python
# melt - wide data to long data ( converting more columns into less cols )

df =  pd.DataFrame({
    "country":["USA","USA", "India", "India"],
    "year":[2020, 2021, 2020, 2021],
    "sales":[92992, 29919, 18282, 19920],
    "profit":[919921, 282919, 828921, 289210],
})

melted_df = df.melt(
    id_vars=["country", "year"], # keep column in new DF
    value_vars=["sales", "profit",], # which column need to merge in new DF
    var_name="metrics", # new variable column name in new DF
    value_name="value", # value column name in new DF
)
melted_df
```

```python
# pivot - long data to wide data
# pivot(index-keep values, columns-demerge columns, values)
melted_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)
```

# Merging and Joining Data

```python
# inner join, outer, Right, Left Join
df_cutsomers=pd.DataFrame({
    "customer_id":[1, 2, 3, 4, 5],
    "name":["Pratik", "John", "Shaw", "Ben", "Jake"]
})

df_orders=pd.DataFrame({
    "order_id":[101, 102, 103, 104],
    "customer_id":[2, 1, 4, 7],
    "amount":[200, 100, 400, 500],
})
```

```python
pd.merge(df_cutsomers, df_orders, on="customer_id") # inner join
pd.merge(df_cutsomers, df_orders, on="customer_id", how='left') 
pd.merge(df_cutsomers, df_orders, on="customer_id", how='right')
pd.merge(df_cutsomers, df_orders, on="customer_id", how='outer') # full join
```

# Data concatenation

```python
# Stacking DataFrames on top of each other (rows) or side by side (columns)
df1 = pd.DataFrame({
    "id":[1, 2, 3],
    "name":["Pratik", "John", "Shaw",]
})

df2 = pd.DataFrame({
    "id":[4, 5, 6],
    "name":[ "Ben", "Jake", "Ross"]
})

# row wise concatenation
pd.concat([df1, df2])
pd.concat([df1, df2], ignore_index=True) # overwrite original lables

# column wise concatenation
pd.concat([df1, df2], axis=1)
```
