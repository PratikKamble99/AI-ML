```python
from bs4 import BeautifulSoup

with open("scraped_data/data1.html") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
# methods
# find -> finds first matching (soup.h3 also in this way you can find one matching)
soup.find("h1") 

# find_all -> finds all matching
all_h3 = soup.find_all("h3") # it return list ie. we can perform all list operations

all_countries = []

for h3 in all_h3:
    name=h3.get_text(strip=True)
    # find_parent -> returns parent tag info
    parent = h3.find_parent()
    # print(parent["class"]) # get attribute value from element

    # find_next -> returns next matching element
    # print(h3.find_next("div").select("span.country-population"))

    # css selector
    # select -> get elements based on css selector
    # select-one() -> get first element based on css selector
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    # print(population)

    # DOM navigation
    # print(h3.parent)
    # print(h3.children)
    # print(h3.descendants)
    # print(h3.next_sibling)
    # print(h3.previous_sibling
    # h3.previous_sibling
    # h3.parent
    print(h3.next_element)
    # h3.previous_element
    
    all_countries.append({
        "name": name,
        "population":population
    })

    # all_countries.append([name, population])

# all_countries
```

```python
import pandas as pd

df = pd.DataFrame(all_countries)
df
```

```python
df.to_csv("cleaned_data/data1.csv", index=False)
```
