# Histogram

A histogram shows the distribution of numerical data by:
- Grouping the data into bins (intervals/intervals) - x axis
- Counting how many values fall under each bin (frequency) - y axis

```python
import numpy as np
np.random.seed(0) # sets the starting sequence
scores = np.random.normal(50, 10, 100) # normal distribution with mean=70, std=10, size=100
```

```python
# bins=how many intervals to divide the data into
plt.hist(scores, edgecolor="black") 
# plt.hist(scores, edgecolor="black", bins=10) 
# plt.hist(scores, edgecolor="black", bins=[0, 20, 40, 60, 80, 100]) 
plt.axvline(x=35, color="black", linestyle="--")
plt.xlabel("Scores")
plt.ylabel("no. of students")
```

```python
# multiple datasets
legit_transactions = np.random.normal(50, 10, 1000)
fraud_transactions = np.random.normal(50, 20, 500)
plt.hist(fraud_transactions, edgecolor="black", color="red", alpha=0.5, bins=10, label="Fraud")
plt.hist(legit_transactions, edgecolor="black", color="green", alpha=0.5, bins=10, label="Legitimate")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.legend()
plt.grid(visible=True)
```

# Box Plot
A Box plot is statistical visualization that summarize the distribution of dataset using five key numbers
- Minimum( lower whisker): The smallest value in the dataset
- First Quartile (Q1): The value below which 25% of the data falls
- Median (Q2): The middle value of the dataset
- Third Quartile (Q3): The value below which 75% of the data falls
- Maximum ( upper whisker): The largest value in the dataset
- Value which is above maximum or below minimum is called outliers. They are represented as individual points in the box plot. eg. 20 in below plot

formulas
1. lower = Q1 - 1.5 * IQR
2. upper = Q3 + 1.5 * IQR
3. IQR(Inner Quartile Range) = Q3 - Q1 

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]
# sorted = [4, 5, 6, 7, 8, 9, 10, 12, 15]
Q3 = np.percentile(data, 75)
Q1 = np.percentile(data, 25)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print("Q1:", Q1, "Q3:", Q3, "IQR:", IQR, "Lower:", lower, "Upper:", upper)
# Note:- you print lower and upper values which are between outliers range from data.

# plt.boxplot(data)
plt.boxplot(data, vert=False)
plt.grid(True)
```

# Subplots

```python
x = [1,2, 3, 4, 5, 6, 7, 8]

y1 = [np.sqrt(i) for i in x]
y2 = [i*2 for i in x]
y3 = [i**2 for i in x]
y4 = [i**3 for i in x]

# plot at index 1
plt.subplot(2, 2, 1) # (rows, columns, index)
plt.plot(x, y1, label="sqrt")

# plot at index 2
plt.subplot(2, 2, 2) # (rows, columns, index)
plt.plot(x, y2, label="linear")

plt.subplot(2, 2, 3) # (rows, columns, index)
plt.plot(x, y3, label="quadratic")

plt.subplot(2, 2, 4) # (rows, columns, index)
plt.plot(x, y4, label="cubic")
```
