```python
import json
```

```python
import os
os.getcwd()
```

```python
# load json
def load_json(filepath):
    with open(filepath) as f:
        data = json.load(f)
    return data
```

```python
data = load_json("./data/data.json")
```

```python
# Clean data
def clean_data(data):
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # Add missing data
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None
            
        # Remove Duplicate
        if(user["first_name"] in unique_users):
            continue

        unique_users.add(user["first_name"])
        cleaned_data.append(user)
        print(user)
    
```

```python
clean_data(data)
```
