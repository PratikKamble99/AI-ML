```python
# load json
def load_json(filepath):
    with open(filepath) as f:
        data = json.load(f)
    return data
```

```python
data = load_json("./data.json")
```

```python
# Clean data
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # Make consistant data
        raw_rating = str(user.get("ratings")).strip().lower()
        if(raw_rating in text_to_num):
            user["ratings"] = text_to_num[raw_rating]
        
        # Add missing data
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

            
        # Remove Duplicate
        if(user["name"] in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
    
```

```python
new_data = clean_data(data)
print(new_data)
```

```python
# GET AVG RATINGS OF ALL USERS
def get_avg_ratings(data):
    total_ratings = 0
    for user in data:
        total_ratings += float(user["ratings"])

    print(f"AVG ratings = {total_ratings/len(data)}")
```

```python
get_avg_ratings(new_data)
```

```python
# CALCULATE % WHO GAVE < 3 RATINGS
def get_pecents_ratings(threashold: int, data):
    below_3_ratings_users = 0
    for user in data:
        if(int(user["ratings"]) < threashold):
            below_3_ratings_users+=1
        else: continue
    print(f"Percents users who gave < 3 ratings: {(below_3_ratings_users/len(data))*100}")
```

```python
get_pecents_ratings(3, new_data)
```

```python
# BRAND RECOMMANDATION BASED ON RATINGS
def recommandations(data):
    recommandations = []
    for user in data:
        current_recommandation = {}

        current_recommandation["name"] = user.get("name")
        current_recommandation["ratings"] = user.get("ratings")

        if(int(user["ratings"]) >= 4):
           current_recommandation["brand"] = user["brand"]
        else:
            current_recommandation["brand"] = "Samsung"

        recommandations.append(current_recommandation)

    print(f"recommandations {recommandations}")
        
```

```python
recommandations(new_data)
```
