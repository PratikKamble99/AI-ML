```python
import numpy as np
import time
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])
print(arr, type(arr))
```

```python
'''
    Vectorization in NumPy is the practice of performing operations on entire arrays at once, 
    rather than using explicit Python for loops to process individual elements. 
    This approach is significantly faster because the underlying operations are executed by highly optimized, pre-compiled C and Fortran code 
'''

size = 1_000_000

arr = list(range(size))
start = time.time()
sq_list = [i**2 for i in arr]
end = time.time()

print(f"Python list Time: {end-start} seconds")

np_arr = np.array(arr)
start = time.time()
sq_list = np_arr ** 2
end = time.time()

print(f"Numpy array Time: {end-start} seconds")
```

```python
# MEMORY
import sys

# Because of extra metadata in python list it occupies more memory that np array
print(f"python list size = {sys.getsizeof(arr)*len(arr)} bytes")

print(f"Numpy array size = {np_arr.nbytes} bytes")
```

```python
# CREATING ARRAY FROM LIST
arr = np.array([1, 2, 3, 4, 5])
print(arr, arr.dtype)

arr2 = np.array([1, 2, 3, 4, 5, "Prime"])
print(arr2, arr2.dtype, arr2.shape)  # numpy array is homogenios array so it converts into same datatype array

arr2D = np.array([[1, 2],[1, 2],[1, 2]])
print(arr2D, arr2D.shape) # shape give you dimansions of array
```

```python
# CREATE FROM SCRACH
arr = np.zeros((2, 3), dtype="int64") # prefill with zero
print(arr, arr.shape)

arr2 = np.ones((2, 3)) # prefill with zero
print(arr2, arr2.shape)

arr3 = np.full((2, 3), 100) # prefill with val
print(arr3, arr3.shape)

'''
    The identity array is a square array with ones on the main diagonal.
    Others zero
'''

arr4 = np.eye(3, dtype="int64") # Identity metrix, dtype used to pass datatype to element
print(arr4, arr4.shape)

arr5 = np.arange(3, 30) # same as range method
print(arr5)

arr6 = np.linspace(1, 10, 4) #start, stop and last arg is size of array to create evenly spaced array
print(arr6)
```

```python
# Properties
arr = np.array([1, 2, 3, 4, 5, 6])

print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.ndim) # get array diamensions

float_arr = np.astype(np.float64)
print(float_arr, float_ar.dtype)
```
